```python
import vizdoom as vzd
import matplotlib.pyplot as plt
import random

# 1. Configuración del juego
game = vzd.DoomGame()
game.load_config(vzd.scenarios_path + "/basic.cfg")
game.set_window_visible(False)
game.set_screen_format(vzd.ScreenFormat.RGB24)
game.set_screen_resolution(vzd.ScreenResolution.RES_640X480)
game.init()

# 2. Definir acciones posibles [DISPARAR, MOVER_IZQ, MOVER_DER]
actions = [[True, False, False], [False, True, False], [False, False, True]]

# 3. Ejecutar un episodio de prueba
game.new_episode()
for _ in range(15):
    game.make_action(random.choice(actions))

# 4. Capturar el frame actual y mostrarlo en Colab
state = game.get_state()
if state is not None:
    screen_buffer = state.screen_buffer
    plt.figure(figsize=(8, 6))
    plt.imshow(screen_buffer)
    plt.axis('off')
    plt.title("Fotograma capturado desde VizDoom en Colab")
    plt.show()

game.close()
```
